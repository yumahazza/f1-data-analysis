```python
import pandas as pd
import requests

BASE = "https://api.openf1.org/v1"

# 1. Cari session_key (misal Azerbaijan GP 2023 - Race)
sess = requests.get(
    f"{BASE}/sessions",
    params={"year": 2023, "country_name": "Azerbaijan", "session_name": "Race"},
).json()
session_key = sess[0]["session_key"]

# 2. Ambil data lap & stint (kompon ban) untuk 1 driver (misal Verstappen = 1)
laps = pd.DataFrame(
    requests.get(f"{BASE}/laps", params={"session_key": session_key, "driver_number": 1}).json()
)
stints = pd.DataFrame(
    requests.get(f"{BASE}/stints", params={"session_key": session_key, "driver_number": 1}).json()
)
laps[["lap_number", "lap_duration", "duration_sector_1", "duration_sector_2", "duration_sector_3"]].head()
```
